# Default notebook

This default notebook is executed using a Lakeflow job as defined in resources/sample_job.job.yml.

```python
# Set default catalog and schema (create if they don't exist)
catalog = dbutils.widgets.get("catalog")
schema = dbutils.widgets.get("schema")
spark.sql(f"USE CATALOG `{catalog}`")
spark.sql(f"CREATE SCHEMA IF NOT EXISTS `{catalog}`.`{schema}`")
spark.sql(f"USE SCHEMA `{schema}`")
```

```python
#from pyspark.sql import SparkSession
#from pyspark.sql import Row

employees=[(101,"Rahul","IT","Mumbai"),
           (102, "Amit", "HR", "Pune"),
           (103, "Priya", "Finance", "Delhi"),
            (104, "Neha", "IT", "Bangalore"),
            (105, "Karan", "Sales", "Mumbai")
           ]
employees_df=spark.createDataFrame(employees,["empid","name","department","City"])

salary_data=[(101,60000),
(102,55000),
(103, 70000),
(104, 65000),
(106, 50000)]

salary_df=spark.createDataFrame(salary_data,["empid","salary"])
employees_df.createOrReplaceTempView("employees")
salary_df.createOrReplaceTempView("salarys")
display(employees_df)
display(salary_df)


#join_df=employees_df.join(salary_df,"empid","left_semi")
#display(join_df)

joins1=spark.sql("""select * from employees a left semi join salarys b on a.empid=b.empid""")
display(joins1)

# Save the join result as a table in the default catalog/schema
joins1.write.mode("overwrite").saveAsTable(`{catalog}`.`{schema}.employee_salary')


```
